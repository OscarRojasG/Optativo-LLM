```python
import json
from pathlib import Path

def fusionar_archivos_json(lista_rutas_archivos, ruta_salida):
    resultado = {}
    
    # Iterar sobre cada ruta de archivo de la lista
    for ruta in lista_rutas_archivos:
        archivo = Path(ruta)
        
        # Verificar si el archivo realmente existe antes de abrirlo
        if archivo.is_file():
            with open(archivo, 'r', encoding='utf-8') as f:
                datos = json.load(f)  # json.load (sin 's') lee directamente del archivo
                resultado |= datos    # Fusiona el diccionario usando el operador |
        else:
            print(f"Advertencia: El archivo {ruta} no existe y será ignorado.")
            
    # Guardar el diccionario final fusionado en el archivo de salida
    with open(ruta_salida, 'w', encoding='utf-8') as f_salida:
        json.dump(resultado, f_salida, indent=4, ensure_ascii=False)
    
    print(f"¡Fusión completada con éxito! Archivo guardado en: {ruta_salida}")

# --- Ejemplo de uso ---

# 1. Define la lista de archivos que quieres fusionar
archivos_a_fusionar = [
    "../data/attributes.json",
    "../data/attributes2.json"
]

# 2. Define el nombre del archivo de salida
archivo_resultado = "../data/json_fusionado.json"

# 3. Ejecutar la función
fusionar_archivos_json(archivos_a_fusionar, archivo_resultado)
```
